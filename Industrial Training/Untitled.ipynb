```python
import pandas as pd
from sklearn.datasets import load_iris

# Load the dataset
iris = load_iris()
```

```python
iris.target
```

```python
train = pd.DataFrame(iris.data, columns=iris.feature_names)
train
```

```python
print(iris.DESCR)
```

```python
train["Name_of_flower"] = iris.target
# train.head(10)
# train.tail(10)
train.sample(10)
```

```python
train[["sepal length (cm)", "petal length (cm)"]].head()
```

```python
train[train["sepal length (cm)"] > 7]
```

```python
train[~(train["sepal length (cm)"] > 7) | (train["petal length (cm)"] > 7)]

train[(train["Name_of_flower"] == 0) & (train["petal width (cm)"] <2)]
```

*Matplotlib*

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# plot: line_graph

plt.figure(figsize=(10,5))
plt.plot(train.index, train["sepal length (cm)"])
plt.title("Line Plot")
plt.xlabel("index")
plt.ylabel("Sepal Length (cm)")
plt.show()
```

```python
# barchart
plt.figure(figsize=(10,4))
train.groupby("Name_of_flower")["petal length (cm)"].mean().plot(kind="bar")

plt.title("Bar Graph")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
plt.show()
```

```python
plt.figure(figsize=(10,4))

plt.scatter(train["sepal length (cm)"], train["petal length (cm)"])

plt.title("Scatter")
# plt.xlabel("Species")
# plt.ylabel("Sepal Length (cm)")
plt.show()
```
